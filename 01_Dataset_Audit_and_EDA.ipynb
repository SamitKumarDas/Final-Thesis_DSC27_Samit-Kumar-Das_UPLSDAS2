# Objective: 
To assess the structure, quality, completeness, and basic characteristics of the biomedical literature corpus prior to applying NLP techniques.

```python
import pandas as pd
df = pd.read_csv("pubmed_medical_affairs_3layer_corpus.csv")
```

```python
missing_values = df.isnull().sum()

missing_values
```

```python
missing_perc = ((df.isnull().sum()) / (len(df))) *100
missing_perc
```

```python
# Check for completely duplicated rows

duplicate_rows = df.duplicated().sum()
duplicate_rows
```

```python
# Check for completely duplicated pmids

duplicate_pmid = df["pmid"].duplicated().sum()
duplicate_pmid
```

```python
# Check for duplicate PMIDs

duplicate_titles = df["title"].duplicated().sum()
duplicate_titles
```

```python
# Investigating the duplicate titles

duplicate_title_records = df[df["title"].duplicated(keep=False)].sort_values("title")

duplicate_title_records[["pmid", "title", "year", "journal", "layer"]].head(20)
```

```python
duplicte_title_records.shape
```

```python
# Check whether records with duplicate titles also have duplicate abstracts

duplicate_title_records["abstract_duplicate"] = duplicate_title_records["abstract"].duplicated(keep=False)

duplicate_title_records[duplicate_title_records["abstract_duplicate"]][["pmid", "title", "year", "journal", "layer"]].head(20)
```

```python
# Count exact duplicate title + abstract combinations

duplicate_title_abstracts = (duplicate_title_records.duplicated(subset = ["title","abstract"]).sum())
duplicate_title_abstracts
```
